# 21 cm Sense Sensitivity and Foreground Testing

## Basic Imports

```python
import matplotlib.pyplot as plt
import numpy as np
from astropy import units as u, constants as c

import py21cmsense as p21s
```

```python
zs = np.linspace(5.5, 13.5, 1024)
fs = (1.42 * u.GHz / zs).to(u.MHz)

aastar100 = p21s.Observatory.from_ska("AA*", frequency=100*u.MHz)

aastar1km = aastar100.clone(max_antpos=1000 * u.m)
```

```python
sensitivity = p21s.PowerSpectrum(
    observation=p21s.Observation(
        observatory=aastar1km
    )
)

power_std = sensitivity.calculate_sensitivity_1d()

sensitivity.plot_sense_1d()
```
